```sql
create table if not exists smartphones
(id int,name string, brand string,yeat int)
```

```sql
insert into smartphones (id, name, brand, yeat) values
(1, 'Galaxy S21', 'Samsung', 2021),
(2, 'iPhone 13', 'Apple', 2021),
(3, 'Pixel 6', 'Google', 2021),
(4, 'OnePlus 9', 'OnePlus', 2021);
```

```sql
show tables
```

```sql
create view view_apple_phones as select * from smartphones where brand = 'Apple';
```

```sql
select * from view_apple_phones;
```
